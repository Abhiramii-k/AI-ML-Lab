```python
import heapq

# Graph
graph = {
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 1)],
    'C': [('D', 2)],
    'D': [('G', 3)],
    'G': []
}

def ucs(graph, start, goal):

    # (cost, node)
    queue = [(0, start)]

    visited = set()

    while queue:

        cost, node = heapq.heappop(queue)

        # Goal found
        if node == goal:
            print("Goal Found")
            print("Total Cost =", cost)
            return

        if node not in visited:

            visited.add(node)

            print("Visited:", node)

            for neighbor, edge_cost in graph[node]:
                heapq.heappush(queue, (cost + edge_cost, neighbor))

# Function Call
ucs(graph, 'A', 'G')
```

```python
import heapq

# Graph
graph = {
    'A': [('B', 2), ('C', 5)],
    'B': [('D', 1)],
    'C': [('D', 2)],
    'D': [('G', 3)],
    'G': []
}

# Heuristic value h(n)
heuristic = {
    'A': 6,
    'B': 4,
    'C': 3,
    'D': 2,
    'G': 0
}


def a_star(start, goal):

    # Priority Queue: (f, node, g)
    # f = g + h
    queue = [(heuristic[start], start, 0)]

    visited = set()

    while queue:

        f, node, g = heapq.heappop(queue)

        print("Visiting:", node)

        # Goal check
        if node == goal:
            print("Goal Found")
            print("Cost =", g)
            return

        if node not in visited:

            visited.add(node)

            # Check neighbors
            for neighbor, cost in graph[node]:

                new_g = g + cost

                # f(n) = g(n) + h(n)
                new_f = new_g + heuristic[neighbor]

                heapq.heappush(
                    queue,
                    (new_f, neighbor, new_g)
                )


# Function call
a_star('A', 'G')
```
