Finally, the dataset was checked again to ensure that no missing values remained.

## Step 5: Removing Duplicate Records

Duplicate records were identified using the `duplicated()` function. The total number of duplicate rows was counted using `sum()`. The `drop_duplicates()` function was then used to remove duplicate records from the dataset. Finally, the dataset was checked again to confirm that no duplicate records remained.

```python
Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

print("Q1 =", Q1)
print("Q3 =", Q3)
IQR = Q3 - Q1

print("IQR =", IQR)
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit =", lower_limit)
print("Upper Limit =", upper_limit)
```

```python
outliers = df[(df["age"] < lower_limit) | (df["age"] > upper_limit)]

outliers
```

```python
print("Number of outliers:", outliers.shape[0])
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
sns.boxplot(x=df["age"])
plt.title("Box Plot of Age")
plt.show()
```

## Step 6: Outlier Detection

Outliers were detected in the `age` column using the Interquartile Range (IQR) method. The first quartile (Q1), third quartile (Q3), and IQR were calculated. Lower and upper limits were determined using the 1.5 × IQR rule, and values outside these limits were identified as outliers. A box plot was created to visualize the distribution and highlight the outliers.

```python
df = df[(df["age"] >= lower_limit) & (df["age"] <= upper_limit)]
```

```python
print("Dataset Shape After Removing Outliers:", df.shape)
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x=df["age"])
plt.title("Box Plot After Removing Outliers")
plt.show()
```
